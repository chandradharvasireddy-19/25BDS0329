```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('/content/airline-passengers.csv')
df.head()
```

```python
df['Month'] = pd.to_datetime(df['Month'])
df.dtypes
```

```python
print(df.columns)
```

```python
df['month'] = pd.to_datetime(df['month'])
df.dtypes
```

```python
df = df.set_index('month')
```

```python
# Check random 5 rows
df.sample(5, random_state=0)
```

```python
# Time-based indexing
df.loc['1955']
```

```python
# Time-based indexing for a specific month
df.loc['1955-10']
```

```python
sns.set(rc={'figure.figsize': (12, 5)})

sns.lineplot(data=df, x=df.index, y='total_passengers')

plt.title('Airline Passengers Time Series')
plt.xlabel('Date')
plt.ylabel('Passengers')
plt.show()
```

```python
df['Cumulative_Passengers'] = df['total_passengers'].cumsum()

df['Cumulative_Passengers'].plot(figsize=(12, 5))

plt.title('Cumulative Sum of Airline Passengers')
plt.xlabel('Date')
plt.ylabel('Cumulative Passengers')
plt.show()
```
